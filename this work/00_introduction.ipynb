# Predicting Persuasive Arguments in Reddit's Change My View

## 1.1. The CMV Forum on Reddit and the "Delta" Concept

[Change My View](https://www.reddit.com/r/changemyview/) (CMV) is a popular subreddit on Reddit where users engage in structured debates with the goal of persuading others to reconsider their viewpoints. Unlike typical internet arguments, CMV encourages respectful discourse, requiring users to provide well-reasoned arguments and evidence. A core feature of CMV is the "delta" system: if a user successfully changes another participant's mind, the persuaded user awards them a "delta" symbol (∆). This delta serves as an indicator of a convincing argument, making CMV a unique platform for studying persuasion in online discussions. The structured format of the forum, combined with the delta-based feedback, provides a valuable dataset for analyzing what makes an argument compelling.

## 1.2. Goals of the Work

The main objective of this project is to explore the properties that differentiate persuasive arguments from non-persuasive ones. Specifically, I aim to:

* Identify textual properties that contribute to persuasion, including the role of comment length.
* Analyze the differences between convincing and non-convincing arguments.
* Develop a machine learning model capable of predicting whether an argument is convincing, and test whether a small set of explicitly-defined rhetorical strategies (e.g. refutation, justification) adds predictive information beyond simpler textual features and comment length.
* Check how well these findings hold up by validating them on an independent dataset of persuasive arguments, built by other researchers using a different methodology.

By achieving these goals, I aim to gain deeper insights into the mechanics of persuasion and contribute to the broader field of computational argumentation analysis.

## 1.3. Motivation

Previous research has leveraged CMV discussions to study persuasion, but earlier studies were often limited by the analytical methods available at the time. With recent advancements in natural language processing (NLP) and machine learning, I now have the tools to extract more nuanced insights from text data. By applying modern text mining techniques, such as transformer-based language models and sophisticated feature engineering methods, I can potentially improve upon previous results and enhance my understanding of what makes an argument compelling. This work has practical applications in areas such as automated debate analysis, content moderation, and persuasive writing assistance.

## 1.4. Process Overview

The work proceeds in five broad stages:

1. **Dataset construction.** Posts and comments are collected from r/changemyview via the Reddit API, then parsed into a structured dataset: one row per comment, labeled by whether it received a delta.

2. **Feature engineering.** A wide range of textual properties are extracted per comment — sentiment, tone, style, readability, rhetorical appeals, evidence use, and document embeddings — alongside the same features computed for the original post.

3. **Initial modeling.** A classifier is trained to predict whether a comment is convincing, evaluated with a thread-grouped split to avoid leakage between training and test data. This stage is also where comment length emerges as an unexpectedly strong predictor, which the later stages account for explicitly.

4. **Rhetorical strategy scoring and analysis.** A small set of rhetorical strategies is defined and scored per comment by an LLM, then tested for whether it adds predictive value beyond comment length and the other engineered features — including after comment length is statistically controlled for.

5. **External validation.** The same scoring method and modeling approach are applied, unchanged, to an independent persuasion dataset (Winning Arguments, built by other researchers from the same subreddit using a different construction methodology) to check how well the findings generalize.
